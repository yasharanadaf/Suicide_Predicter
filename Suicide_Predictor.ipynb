```python
import joblib
from customtkinter import *

tfidf = joblib.load('tfidf_vectorizer.pkl')
clf = joblib.load('logistic_model.pkl')
label_encoder = joblib.load('label_encoder.pkl')
print("Models and vectorizer loaded successfully.")

def clean_text(text):
    return ''.join(char.lower() for char in text if char.isalnum() or char.isspace())

def predict():
    user_input_text = entry.get().strip()
    if not user_input_text:
        result.configure(text="⚠ Please enter some text", font=("Arial Rounded MT", 25))
        return

    cleaned_input = clean_text(user_input_text)
    vectorized_input = tfidf.transform([cleaned_input])
    prediction_index = clf.predict(vectorized_input)[0]
    predicted_class_name = label_encoder.inverse_transform([prediction_index])[0]

    if predicted_class_name == 'suicide':
        result.configure(text=(f"❗ {predicted_class_name} ❗\n""You might be feeling low 😞\n""We detected signs of distress ❗\n""You're not alone. Help is available 💗"),
            font=("Arial Rounded MT", 25))
    else:
        result.configure(text=f"❗ {predicted_class_name} ❗\nYou seem okay 😊",
                         font=("Arial Rounded MT", 25))


def clear():
    entry.delete(0, END)
    result.configure(text="")

#GUI 
window = CTk()
window.title('SUICIDE PREDICTOR')
window.geometry('800x680')
set_appearance_mode('dark')
  
# Header
header = CTkLabel(window, text="SUICIDE PREDICTOR 📧",font=("Arial Rounded MT Bold", 50),text_color="#3B3B98")
header.pack(pady=10)

msg1 = CTkLabel(window, text="Your words matter. We're listening",font=("Arial Rounded MT", 18),text_color="#3B3B98")
msg1.pack(pady=5)

label=CTkLabel(window, text="🧠 Let's Talk -\n ✍ How are you feeling today ?",font=("Arial Rounded MT", 25),text_color="#D2C1B6")
label.pack(pady=20)

#entry
entry = CTkEntry(window, width=600, height=150, font=("Helvetica", 20),placeholder_text="Type how you feel today...")
entry.pack(pady=10)

#buttons
predict_btn = CTkButton(window, text='PREDICT',font=('Arial Rounded MT Bold', 25),command=predict,fg_color='#3B3B98', text_color='#D2C1B6')
predict_btn.pack(pady=35)
predict_btn .place(x=150,y=400)


clear_btn = CTkButton(window, text='CLEAR',font=('Arial Rounded MT Bold', 25),command=clear,fg_color='#3B3B98', text_color='#D2C1B6')
clear_btn.pack(pady=35)
clear_btn .place(x=500,y=400)


#result 
result = CTkLabel(window, text='',font=('Arial Rounded MT Bold', 40),text_color='#D2C1B6')
result.pack(pady=75)

#footer
footer = CTkLabel(window,text="💡 Developed with care by @Yashara | Not a medical tool",font=("Arial", 14),text_color="gray")
footer.pack(side="bottom", pady=10)

window.mainloop()
```
